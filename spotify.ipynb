```python
sptfy_df.dropna(subset=['track_name', 'track_artist', 'track_album_name'], inplace=True)
sptfy_df.isnull().sum()
```

```python
sptfy_df['duration_min'] = sptfy_df['duration_ms'] / 60000
sptfy_df.drop('duration_ms', axis=1, inplace=True)
```
